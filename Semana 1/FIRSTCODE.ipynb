```python
import random ## Importamos un biblioteca que nos otorga python, la cual tiene algunas funcionalidades que crean numeros aleatorios para nosotros

numero_secreto = random.randint(1, 10) ## Crea un numero aleatorio entre 1 a 10 y pues se la asigna a numero_secreto

intentos = 3

print("Estoy pensando en un numero entre el 1 y el 10")

while intentos > 0:

  posible = int(input("Adivina el numero que escogi aleatoriamente: \n"))
  print("Tienes", intentos, "intentos")

  if posible == numero_secreto:
    print("Felicidades, adivinaste el numero que escogi!!")
    break

  elif posible < numero_secreto:
    print("El numero que escogi es mayor, intentalo de nuevo!!")

  else:
    print("El numero que escogi es menor, intentalo de nuevo")
  intentos -= 1

if intentos == 0:
  print("Lo siento, te quedaste sin intentos :(, el numero que escogi era: ", numero_secreto)

```
