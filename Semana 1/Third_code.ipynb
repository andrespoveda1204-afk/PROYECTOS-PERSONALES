```python
num1 = int(input("Ingresa un numero: "))
num2 = int(input("Ingresa otro numero: "))

print(f"La suma entre el numero {num1} y {num2} es {num1 + num2}")
```

```python
## CALCULADORA
print("\n--- BIENVENIDO A MI CALCULADORA ---\n")
num1 = int(input("Ingresa un numero: "))
num2 = int(input("\nIngresa otro numero: \n"))
div = num1 / num2 if num2 != 0 else "No existe la division entre cero"
print(f"--- OPERACIONES MATEMATICAS BASICAS CON LOS NUMEROS {num1} y {num2} ---")
print(f"Suma de {num1} y {num2}: {num1 + num2}")
print(f"\nResta de {num1} y {num2}: {num1 - num2}")
print(f"\nDivision de {num1} y {num2}: {div}")


```
