```python
## Las funciones son easys pero es importante recordar la importancia y la utilidad tan grande al llamar funciones y pasarle datos a los parametros de estas.
def saludo (name):
  print(f"Hola {name}, ¿como estas ? ")

nombre = input("¿What is your name?: ")
saludo (nombre)
```

```python
def suma (a,b):
  print(f"La suma de {a} y {b} es {a + b}")

suma (3, 4)
```

```python
## DECLARACIONES DE RETORNO
def mult (a,b):
  return a * b ## Basicamente, reemplazo por print, y con return se puede realizar operaciones matematicas directamente

resultado = mult (2,4)
print(f"La multiplicacion de 2 y 4 es {resultado}")
```

```python
## PROYECTO DIA 6 -> Quiz de matematicas
import random
def preguntas ():
  num1 = random.randint (1, 10)
  num2 = random.randint (1, 10)
  operador = random.choice (["+", "-", "*"])

  if operador == "+":
    resultado = num1 + num2
  elif operador == "-":
    resultado = num1 - num2
  else:
    resultado = num1 * num2
  return f"{num1} {operador} {num2}", resultado

preguntas()

def quiz ():
  puntaje = 0
  rondas = 5
  print("--- BIENVENIDO A TU QUIZ DE MATEMATICAS ---\n")
  print("Habran preguntas basicas de matematicas y tendras que responder correctamente")
  for i in range (rondas):
    pregunta, correcta = preguntas () ## a pregunta se le asignar la cadena que se retorno antes de la coma en la funcion preguntas ({num1} {operador} {num2}) y correcta tendra el resultado que se retorno en la funcion preguntas
    print(f"--- Pregunta {i + 1}: ¿cuanto es {pregunta}? ---\n")
    answer = int(input(""))
    if answer == correcta :
      print("\nFelicidades!! respondiste correctamente\n")
      puntaje += 20
    else:
      print(f"\nLo siento ;c, {pregunta} es {correcta}, no {answer}\n")
  print("\nEl puntaje total es 100, cada pregunta vale 20 puntos\n")
  if puntaje >= 0 and puntaje <= 40:
    print (f"Tu puntaje fue de {puntaje}, tienes que mejorar!!!")
  elif puntaje == 60:
    print(f"Tu puntaje fue de {puntaje}, regular pero bien")
  elif puntaje == 80:
    print(f"Tu puntaje fue de {puntaje}, felicidades!!, bastante bien")
  else:
    print(f"Excelente, tuviste un perfecto y hermoso puntaje de {puntaje}, enseñame matematicas jajaj")
quiz ()
```
