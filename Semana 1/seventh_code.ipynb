```python
## LISTAS
lista_de_compras = ["arroz", "pan", "fruta"]
print(lista_de_compras)
## ACCEDER A ELEMENTOS DE UNA LISTA
print( lista_de_compras [0]) ## Acceder a elementos de la lista directamente, en la posicion 0 esta arroz, con corchetes

## AÑADIR ELEMENTOS A UNA LISTA
lista_de_compras.append("carne") ## append es para agregar un elemento al final de la lista
lista_de_compras.insert(1, "Chocolate") ## insert es para agregar un elemento a una lista en una posicion especifica, en este caso en el indice(posicion) 1
print(lista_de_compras)

lista_de_compras.remove("carne") ## Remove es para eliminar un elemento especifico de una lista, en este caso, carne

print(lista_de_compras)

lista_de_compras.pop(0) ## Pop sin indice es para eliminar el ultimo elemento en una lista

print(lista_de_compras)

## RECORRER UNA LISTA:
#for i in lista_de_compras:
  #print(f"- {i}")

## RECORRER UNA LISTA ENNUMERADA (PARA MAS ELEGANCIA)
for index, i in enumerate(lista_de_compras):
  print(f"{index + 1}. {i}")

## MAS METODOS DE LISTAS
```

```python
## MAS METODOS DE LISTAS
lista_de_compras = ["Arroz", "Pan", "Fruta", "Pechuga", "Platano", "Carne"]
lista_de_compras.sort() ## Sort ordena los elementos de la lista en orden alfabetico

print(lista_de_compras)

lista_de_compras.reverse() ## Reverse para invertir el orden de la lista

print(lista_de_compras)

#lista_de_compras.clear() ## Clear limpia todos los elementos que esten en una lista

print(lista_de_compras[0], "Fue removido")
```

```python
## PROYECTO CARRITO DE COMPRAS
carrito_compras = []
def menu ():
  print("\n--- MENU DE LISTA DE COMPRAS ---\n")
  print("1. Mirar el carrito de compras")
  print("\n2. Agregar una compra al carrito de compras\n")
  print("3. Eliminar una compra del carrito de compras\n")
  print("4. Limpiar las compras del carrito de compras\n")
  print("5. Salir")

while True:
  menu ()
  opcion = input("Selecciona la opcion que desees (1-5): ")
  if opcion == "1":
    print("\n--- CARRITO DE COMPRAS ---\n")
    if not carrito_compras:
      print("Tu carrito de compras esta vacio\n")
    else:
      for index, i in enumerate(carrito_compras):
        print(f"{index + 1}. {i}")
  elif opcion == "2":
    compra = input("Ingresa la compra que quieres añadir: ")
    carrito_compras.append (compra)
    print(f"\n{compra} fue añadida a tu carrito de compras")
  elif opcion == "3":
    print("\n--- CARRITO DE COMPRAS ---\n")
    if not carrito_compras:
      print("Tu carrito de compras esta vacio, no hay nada que eliminar\n")
    else:
      for index, i in enumerate(carrito_compras):
        print(f"{index + 1}. {i}")
      delete = input("\nEstas son tus compras, ¿cual compra deseas eliminar?: ")
      if delete in carrito_compras:
        carrito_compras.remove(delete)
        print(f"\n{delete} fue removido de tu carrito de compras\n")
      else:
        print(f"{delete} no existe en tu carrito de compras")
        pass
  elif opcion == "4":
    carrito_compras.clear()
    print("Tu carrito de compras fue limpiado correctamente")
  elif opcion == "5":
    print("\nASI QUEDO TU CARRITO DE COMPRAS:\n")
    for index, i in enumerate(carrito_compras):
        print(f"{index + 1}. {i}")
    print("\nVuelve pronto, feliz dia !!")
    break

  else :
    print("\n Esa opcion no esta dentro del menu, intentalo otra vez!!")





```
