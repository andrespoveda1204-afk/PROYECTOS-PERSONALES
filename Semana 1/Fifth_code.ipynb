```python
## BUCLES FOR
# Estructura del for :
for variable in range (inicio, fin, paso) #Inicio: Desde que indice va a iniciar nuestra variable (el valor determinado siempre es cero), variable, Fin: Hasta donde va a recorrer nuestra variable, Paso: el incremento de nuestra variable por cada ciclo realizado (el valor determinado siempre es 1)
## Codigo a repetir
```

```python
for i in range (5):
  print(i)
```

```python
for i in range (5, 0, -1):
  print (i)
```

```python
## BUCLES WHILE
#while condicion(verdadera):
  # Codigo a repetir

count = 5
while count > 0:
  print(count)
  count -= 1
```

```python
## uso de time.sleep() para pausas
import time ## aqui es donde existe la funcion sleep, modulo disponible en python

for i in range (5, 0, -1):
  print (i)
  time.sleep(1)
print("Feliz año nuevo")
```

```python
## Proyecto: Temporizador con un numero inicial dada por el usuario y hacer una cuenta regresiva hasta cero con un retraso de 1 segundo
import time

inicio = int(input("Ingresa un numero inicial para la cuenta regresiva: "))
for i in range (inicio, -1, -1):
  print(i)
  time.sleep(1)
print("SEEEEXOOOOOO")
```
