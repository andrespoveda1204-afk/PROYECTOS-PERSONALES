```python
num = 4

if num > 5:
  print(f"{num} es mayor que 5")
elif num == 5:
  print(f"{num} es igual a 5")
else:
  print(f"{num} es menor que 5")
```

```python
## COMPARACION ENTRE DOS NUMEROS
num1 = int(input("Ingresa un numero: "))
num2 = int(input("\nIngresa otro numero: \n"))

if num1 > 0 and num2 >0:


  if num1 == num2:
    print(f"Los numeros que elegiste ({num1} y {num2}) son iguales")

  elif num1 > num2:
    print(f"El numero {num1} es mayor que el numero {num2}")

  elif num2 > num1:
    print (f"El numero {num2} es mayor que el numero {num1}")

  print("Ningun numero escogido es cero")

else:
  print("Alguno de los numeros es cero")


```
